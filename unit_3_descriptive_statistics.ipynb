```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import skew

# 1. Generate a sample dataset of house prices
# Let's create a synthetic dataset that is slightly skewed to the right
np.random.seed(42) # for reproducibility
house_prices = np.concatenate([
    np.random.normal(loc=250000, scale=50000, size=700), # Majority of houses
    np.random.normal(loc=600000, scale=100000, size=200), # More expensive houses
    np.random.normal(loc=150000, scale=30000, size=100) # Less expensive houses
])

# Ensure all prices are positive
house_prices = house_prices[house_prices > 0]

# Create a Pandas Series for easy handling
house_prices_series = pd.Series(house_prices, name='House Prices')

print(f"Generated {len(house_prices_series)} sample house prices.")
print(f"Minimum Price: ${house_prices_series.min():,.2f}")
print(f"Maximum Price: ${house_prices_series.max():,.2f}")
print(f"Mean Price: ${house_prices_series.mean():,.2f}\n")

# 2. Plot the histogram of house prices
plt.figure(figsize=(10, 6))
sns.histplot(house_prices_series, bins=50, kde=True, color='skyblue', edgecolor='black')
plt.title('Distribution of House Prices')
plt.xlabel('House Price ($)')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

# 3. Calculate skewness
skewness_value = skew(house_prices_series)
print(f"\nSkewness of House Prices: {skewness_value:.4f}")
```

```python
import pandas as pd
import numpy as np

# 1. Create a sample dataset of student marks
# Let's include some potential outliers for demonstration
marks = [
    65, 70, 72, 75, 78, 80, 81, 83, 85, 88,
    90, 92, 95, 98, 100, # Regular marks
    30, 45, # Low outliers
    105, 110 # High outliers (e.g., bonus points, data entry errors)
]

marks_series = pd.Series(marks, name='Student Marks')

print("### Student Marks Data ###\n")
print(marks_series)
print("\n")

# 2. Calculate the Interquartile Range (IQR)
Q1 = marks_series.quantile(0.25)
Q3 = marks_series.quantile(0.75)
IQR = Q3 - Q1

print(f"First Quartile (Q1): {Q1}")
print(f"Third Quartile (Q3): {Q3}")
print(f"Interquartile Range (IQR): {IQR}\n")

# 3. Define outlier bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Lower Bound for Outliers: {lower_bound}")
print(f"Upper Bound for Outliers: {upper_bound}\n")

# 4. Detect outliers
outliers = marks_series[(marks_series < lower_bound) | (marks_series > upper_bound)]

if not outliers.empty:
    print("### Detected Outliers ###\n")
    print(outliers)
else:
    print("No outliers detected in the student marks using the IQR method.")

# Optional: Visualize with a box plot to confirm
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 6))
sns.boxplot(y=marks_series, palette='coolwarm')
plt.title('Box Plot of Student Marks with Outliers')
plt.ylabel('Marks')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### Removing Outliers from Sales Data using the IQR Method

This section demonstrates how to identify and remove outliers from a sample sales dataset using the Interquartile Range (IQR) method. Outliers can significantly affect statistical analysis and model training, so their removal or treatment is often a necessary data preprocessing step.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Create a sample sales dataset with some outliers
np.random.seed(42) # for reproducibility
sales_data = np.concatenate([
    np.random.normal(loc=1000, scale=200, size=90), # Regular sales
    np.random.normal(loc=5000, scale=500, size=5),  # High outliers
    np.random.normal(loc=100, scale=30, size=5)   # Low outliers
])

sales_series = pd.Series(sales_data, name='Daily Sales ($)')

print("### Original Sales Data (with potential outliers) ###\n")
display(sales_series.describe())

# Visualize original data
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.histplot(sales_series, bins=30, kde=True, color='purple', edgecolor='black')
plt.title('Original Sales Distribution')
plt.xlabel('Sales Amount ($)')
plt.ylabel('Frequency')

plt.subplot(1, 2, 2)
sns.boxplot(y=sales_series, color='purple') # Changed palette to color
plt.title('Original Sales Box Plot')
plt.ylabel('Sales Amount ($)')
plt.tight_layout()
plt.show()

# 2. Calculate the Interquartile Range (IQR) for sales data
Q1_sales = sales_series.quantile(0.25)
Q3_sales = sales_series.quantile(0.75)
IQR_sales = Q3_sales - Q1_sales

print(f"\nFirst Quartile (Q1): {Q1_sales:.2f}")
print(f"Third Quartile (Q3): {Q3_sales:.2f}")
print(f"Interquartile Range (IQR): {IQR_sales:.2f}\n")

# 3. Define outlier bounds
lower_bound_sales = Q1_sales - 1.5 * IQR_sales
upper_bound_sales = Q3_sales + 1.5 * IQR_sales

print(f"Lower Bound for Outliers: {lower_bound_sales:.2f}")
print(f"Upper Bound for Outliers: {upper_bound_sales:.2f}\n")

# 4. Identify outliers
outliers_sales = sales_series[
    (sales_series < lower_bound_sales) |
    (sales_series > upper_bound_sales)
]

print("### Detected Outliers in Sales Data ###\n")
if not outliers_sales.empty:
    display(outliers_sales)
else:
    print("No outliers detected in the sales data using the IQR method.")

# 5. Remove outliers from the dataset
cleaned_sales_series = sales_series[
    (sales_series >= lower_bound_sales) &
    (sales_series <= upper_bound_sales)
]

print("\n### Cleaned Sales Data (Outliers Removed) ###\n")
display(cleaned_sales_series.describe())

# Visualize cleaned data
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.histplot(cleaned_sales_series, bins=30, kde=True, color='green', edgecolor='black')
plt.title('Cleaned Sales Distribution')
plt.xlabel('Sales Amount ($)')
plt.ylabel('Frequency')

plt.subplot(1, 2, 2)
sns.boxplot(y=cleaned_sales_series, color='green') # Changed palette to color
plt.title('Cleaned Sales Box Plot')
plt.ylabel('Sales Amount ($)')
plt.tight_layout()
plt.show()
```

## Exploratory Data Analysis (EDA) on Iris Dataset

### 1. Load the Iris Dataset

```python
from sklearn.datasets import load_iris
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Load the Iris dataset
iris = load_iris()
iris_df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
iris_df['species'] = iris.target_names[iris.target]

print("Iris Dataset loaded successfully. Displaying the first 5 rows:")
display(iris_df.head())
```

### 2. Display Dataset Information

```python
print("\nDataset Information:")
iris_df.info()
```

### 3. Finding Missing Values

```python
print("\nMissing Values (per column):")
display(iris_df.isnull().sum())
```

### 4. Computing Descriptive Statistics

```python
print("\nDescriptive Statistics:")
display(iris_df.describe())
```

### 5. Plotting Histograms

```python
plt.figure(figsize=(15, 10))
for i, col in enumerate(iris_df.columns[:-1]): # Exclude 'species' column
    plt.subplot(2, 2, i + 1)
    sns.histplot(iris_df[col], kde=True, bins=20, color='skyblue')
    plt.title(f'Histogram of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

### 6. Plotting Box Plots

```python
plt.figure(figsize=(15, 10))
for i, col in enumerate(iris_df.columns[:-1]): # Exclude 'species' column
    plt.subplot(2, 2, i + 1)
    sns.boxplot(y=iris_df[col], color='lightgreen')
    plt.title(f'Box Plot of {col}')
    plt.ylabel(col)
plt.tight_layout()
plt.show()
```

### 7. Detecting and Removing Outliers (using IQR method)

```python
iris_cleaned_df = iris_df.copy()
outlier_indices = pd.Index([])

print("\nDetecting and removing outliers for each numerical column:")
for col in iris_df.columns[:-1]: # Exclude 'species' column
    Q1 = iris_df[col].quantile(0.25)
    Q3 = iris_df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Identify outliers for the current column
    col_outliers = iris_df[(iris_df[col] < lower_bound) | (iris_df[col] > upper_bound)].index

    if not col_outliers.empty:
        print(f"  - Outliers found in '{col}': {iris_df.loc[col_outliers, col].tolist()}")
        outlier_indices = outlier_indices.union(col_outliers)
    else:
        print(f"  - No outliers found in '{col}'.")

if not outlier_indices.empty:
    print(f"\nTotal outliers detected across numerical columns: {len(outlier_indices)} rows.")
    print(f"Rows with outliers (indices): {outlier_indices.tolist()}")
    iris_cleaned_df = iris_df.drop(outlier_indices)
    print("Outliers removed. Displaying head of cleaned dataset:")
    display(iris_cleaned_df.head())
    print("\nShape of original dataset: ", iris_df.shape)
    print("Shape of cleaned dataset: ", iris_cleaned_df.shape)
else:
    print("\nNo outliers were detected in any numerical column, so no rows were removed.")
    iris_cleaned_df = iris_df.copy()
```

### 8. Saving the Cleaned Dataset

```python
output_filename = 'iris_cleaned_dataset.csv'
iris_cleaned_df.to_csv(output_filename, index=False)
print(f"Cleaned Iris dataset saved to '{output_filename}'.")
```
